```python
# ==========================================
# 4. IF CONDITION
# ==========================================

# 4.1 BMI Calculator Function
def check_bmi(height, weight):
    bmi = weight / (height ** 2)
    if bmi >= 30:
        return "Obesity"
    elif 25 <= bmi < 30:
        return "Overweight"
    elif 18.5 <= bmi < 25:
        return "Normal"
    else:
        return "Underweight"


# Country & City Datasets
Australia = ["Sydney", "Melbourne", "Brisbane", "Perth"]
UAE = ["Dubai", "Abu Dhabi", "Sharjah", "Ajman"]
India = ["Mumbai", "Bangalore", "Chennai", "Delhi"]


# 4.2 City Country Lookup Function
def get_country_of_city(city):
    if city in Australia:
        return f"{city} is in Australia"
    elif city in UAE:
        return f"{city} is in UAE"
    elif city in India:
        return f"{city} is in India"
    else:
        return f"{city} is not in database"


# 4.3 Check If Two Cities Belong to Same Country Function
def check_same_country(city1, city2):
    def find_country(city):
        if city in Australia:
            return "Australia"
        if city in UAE:
            return "UAE"
        if city in India:
            return "India"
        return None

    c1 = find_country(city1)
    c2 = find_country(city2)

    if c1 and c2 and c1 == c2:
        return f"Both cities are in {c1}"
    else:
        return "They don't belong to the same country"


# Executing and printing all required example outputs to screen
print("--- TASK 4 OUTPUTS ---")
print("4.1 BMI Category (Height: 1.75m, Weight: 70kg):", check_bmi(1.75, 70))
print("4.2 Single City Check:", get_country_of_city("Abu Dhabi"))
print("4.3 Same Country Check Example 1:", check_same_country("Mumbai", "Chennai"))
print("4.3 Same Country Check Example 2:", check_same_country("Sydney", "Dubai"))
```
