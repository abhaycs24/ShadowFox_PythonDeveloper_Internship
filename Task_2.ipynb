```python
# ==========================================
# TASK 2.1: WEB SCRAPER
# ==========================================
import requests
from bs4 import BeautifulSoup
import pandas as pd

def scrape_quotes():
    print("-- - 1. WEB SCRAPER DEMO ---")
    url = "http://quotes.toscrape.com"
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        quotes_data = []

        # Finding all quote blocks on the page
        quote_elements = soup.find_all('div', class_='quote')

        for q in quote_elements[:5]:  # Extract top 5 quotes for demo
            text = q.find('span', class_='text').get_text()
            author = q.find('small', class_='author').get_text()
            tags = [tag.get_text() for tag in q.find_all('a', class_='tag')]

            quotes_data.append({
                "Quote": text,
                "Author": author,
                "Tags": ", ".join(tags)
            })

        # Display as DataFrame
        df = pd.DataFrame(quotes_data)
        print("Successfully scraped quotes from website:\n")
        print(df.to_string(index=False))
    else:
        print(f"Failed to retrieve web page. Status code: {response.status_code}")

# Run the web scraper
scrape_quotes()

print("\n" + "=" * 60 + "\n")

# ==========================================
# TASK 2.2: HANGMAN GAME
# ==========================================
import random

HANGMAN_PICS = [
    r'''
           +---+
           |   |
               |
               |
               |
               |
        =========''', r'''
           +---+
           |   |
           O   |
               |
               |
               |
        =========''', r'''
           +---+
           |   |
           O   |
           |   |
               |
               |
        =========''', r'''
           +---+
           |   |
           O   |
          /|   |
               |
               |
        =========''', r'''
           +---+
           |   |
           O   |
          /|\  |
               |
               |
        =========''', r'''
           +---+
           |   |
           O   |
          /|\  |
          /    |
               |
        =========''', r'''
           +---+
           |   |
           O   |
          /|\  |
          / \  |
               |
        ========='''
]

WORDS_WITH_HINTS = {
    "python": "A popular programming language",
    "developer": "A person who writes computer code",
    "shadowfox": "The organization running this internship program",
    "algorithm": "A step-by-step procedure for solving a problem",
    "colab": "Google's cloud-based Jupyter notebook service"
}

def play_hangman(simulated_inputs=None):
    print("-- - 2. HANGMAN GAME ---")
    word, hint = random.choice(list(WORDS_WITH_HINTS.items()))
    guessed_letters = set()
    attempts_left = len(HANGMAN_PICS) - 1
    input_idx = 0

    print("Welcome to Hangman!")
    print(f"HINT: {hint}\n")

    while attempts_left > 0:
        print(HANGMAN_PICS[len(HANGMAN_PICS) - 1 - attempts_left])

        # Display word progress (e.g., p _ t h o n)
        word_progress = [letter if letter in guessed_letters else '_' for letter in word]
        print("Word: " + " ".join(word_progress))
        print(f"Guessed letters: {', '.join(sorted(guessed_letters))}")
        print(f"Attempts left: {attempts_left}")

        # Win check
        if '_' not in word_progress:
            print(f"\n🎉 Congratulations! You guessed the word: '{word}'!")
            return

        # Handle input (supports both automated test inputs and manual interactive inputs)
        if simulated_inputs and input_idx < len(simulated_inputs):
            guess = simulated_inputs[input_idx].lower()
            input_idx += 1
            print(f"Guessing letter: {guess}")
        else:
            guess = input("Enter a letter: ").lower()

        if len(guess) != 1 or not guess.isalpha():
            print("⚠️ Please enter a single valid letter.\n")
            continue

        if guess in guessed_letters:
            print(f"⚠️ You already guessed '{guess}'. Try another letter.\n")
            continue

        guessed_letters.add(guess)

        if guess in word:
            print(f"✅ Good job! '{guess}' is in the word.\n")
        else:
            print(f"❌ Sorry, '{guess}' is not in the word.\n")
            attempts_left -= 1

    # Loss check
    print(HANGMAN_PICS[-1])
    print(f"\n💥 Game Over! The word was: '{word}'")

# Run simulated game to demonstrate completion without getting stuck on interactive prompts
sample_guesses = ['a', 'e', 'i', 'o', 'u', 'p', 'y', 't', 'h', 'n', 'd', 'v', 'l', 'r', 's', 'c', 'b']
play_hangman(simulated_inputs=sample_guesses)
```
